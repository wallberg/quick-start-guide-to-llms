```python
import chromadb

# 1. Initialize a persistent local client
client = chromadb.PersistentClient(path="../data/my_chroma_db")
client.get_version()
```

```python
# 2. Create or load a collection
collection = client.get_or_create_collection(name="knowledge_base")
collection.configuration_json
```

```python
# 3. Add documents (Chroma automatically generates embeddings)
collection.add(
    documents=[
        "ChromaDB is an open-source vector database for AI.",
        "Vector databases enable semantic search by meaning."
    ],
    metadatas=[{"source": "docs"}, {"source": "blog"}],
    ids=["doc1", "doc2"]
)
collection.count()
```

```python
# 4. Query by semantic similarity
results = collection.query(
    query_texts=["How does semantic search work?"],
    n_results=1
)
results
```

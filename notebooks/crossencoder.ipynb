```python
from sentence_transformers import CrossEncoder

model = CrossEncoder('cross-encoder/ms-marco-MiniLM-L12-v2')
```

```python
question = "How many people live in Berlin?"
answers = (
    "Berlin had a population of 3,520,031 registered inhabitants in an area of 891.82 square kilometers.",
    "12",
    "1 person lives in Berlin",
    "12 people live in Berlin",
    "3,520,031 people live in Berlin",
    "3,520,031",
    "3,520,031 dogs live in Berlin",
    "1 dog lives in Berlin",
    "No dogs live in Berlin",
    "1,003,520,031 people live in Berlin",
    "No one lives in Berlin",
    "12 people live in Berlitz",
    "12 people live in Barley",
    "Berlin is well known for its museums.",
)

scores = model.predict([(question, answer) for answer in answers])
for score, answer in sorted(zip(scores, answers), key=lambda x: x[0]):
    print(score, answer)
# [ 9.218911  -4.0780287]
```

```python
import torch
from transformers import AutoModelForSequenceClassification, AutoTokenizer

model = AutoModelForSequenceClassification.from_pretrained('cross-encoder/ms-marco-MiniLM-L12-v2')
tokenizer = AutoTokenizer.from_pretrained('cross-encoder/ms-marco-MiniLM-L12-v2')
```

```python
features = tokenizer(['How many people live in Berlin?', 'How many people live in Berlin?'], ['Berlin has a population of 3,520,031 registered inhabitants in an area of 891.82 square kilometers.', 'New York City is famous for the Metropolitan Museum of Art.'],  padding=True, truncation=True, return_tensors="pt")

model.eval()
with torch.no_grad():
    scores = model(**features).logits
    print(scores)
```
